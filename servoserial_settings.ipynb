```python
from servoserial import ServoSerial
import time

servo = ServoSerial()

def write_packet(sid, instruction, params):
    length = len(params) + 2
    checksum = (~(sid + length + instruction + sum(params))) & 0xff
    servo.ser.write(bytes([0xff, 0xff, sid, length, instruction] + params + [checksum]))
    time.sleep(0.1)

def write_reg(sid, addr, *values):
    write_packet(sid, 0x03, [addr] + list(values))

def unlock_eprom(sid):
    write_reg(sid, 0x30, 0x00)

def lock_eprom(sid):
    write_reg(sid, 0x30, 0x01)

def set_servo_id(old_id, new_id):
    unlock_eprom(old_id)
    write_reg(old_id, 0x05, new_id)  # ID register = 0x05
    time.sleep(0.2)
    lock_eprom(new_id)
```

```python
# Servo A만 연결
set_servo_id(0xFE, 1)

```

```python
servo.Servo_serial_control(1, 800)
```

```python
# 전원 끄고 Servo A 분리, Servo B만 연결
set_servo_id(0xFE, 2)

```

```python
servo.Servo_serial_control(2, 3000) # 900 ~ 3000
```

```python
# 전원 끄고 둘 다 연결한 뒤 테스트
servo.Servo_serial_control(1, 800)
time.sleep(0.5)
servo.Servo_serial_control(2, 1000)
```
